## Parte 1: Criação dos Parâmetros

```python
catalog = "movies"

bronze_schema_name = "bronze"

silver_schema_name = "silver"

gold_schema_name = "gold"


bronze_schema = f"{catalog}.{bronze_schema_name}"
silver_schema = f"{catalog}.{silver_schema_name}"
gold_schema = f"{catalog}.{gold_schema_name}"


landing_path = f"/Volumes/{catalog}/{bronze_schema_name}/Landing"
```

## Parte 2: Criar catalogs, schemas e volumes

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS `{catalog}`")
spark.sql(f"USE CATALOG `{catalog}`")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{bronze_schema_name}`")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{silver_schema_name}`")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{gold_schema_name}`")
spark.sql(f"USE SCHEMA `{bronze_schema_name}`")
spark.sql(f"CREATE VOLUME IF NOT EXISTS `Landing`")
```
