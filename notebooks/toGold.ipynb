# 03 - Modelagem Dimensional (Gold) e Desafio de Analytics

Modelagem dimensional da camada Gold: tabelas dimensão, bridges, fato e análise de perguntas de negócio sobre receita, popularidade e lucro.

## 1. Configuração

Imports de funções, tipos e criação do schema `gold`.

```python
from pyspark.sql.functions import (
    col, monotonically_increasing_id, sha2, concat_ws, avg, count, round as spark_round,
    dense_rank, rank, max as spark_max, add_months
)
from pyspark.sql.window import Window
from pyspark.sql.types import LongType, IntegerType, DoubleType, DecimalType

spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

## 2. Tabelas Dimensão

### 2.1 Dimensão Filmes

Cria a dimensão de filmes a partir de `silver.tb_info_filmes`.

```python
df_info = spark.table("silver.tb_info_filmes")

df_dim_movies = (
    df_info
    .withColumn("sk_movie_id", monotonically_increasing_id().cast(LongType()))
    .select(
        "sk_movie_id",
        "id_filme",
        "titulo",
        "data_lancamento",
        "ano_lancamento",
        "duracao_minutos",
        "idioma_original",
        "status_filme",
        "sinopse"
    )
)
df_dim_movies.write.format("delta").mode("overwrite").saveAsTable("gold.dim_movies")

df_movies_lookup = spark.table("gold.dim_movies").select("id_filme", "sk_movie_id")
```

### 2.2 Dimensão Gêneros e Bridge

Cria a dimensão de gêneros e a tabela bridge que relaciona filmes a gêneros.

```python
df_gen = spark.table("silver.tb_generos")

df_dim_genres = (
    df_gen.select("nome_genero").distinct()
    .withColumn("sk_genre_id", monotonically_increasing_id().cast(LongType()))
    .select("sk_genre_id", "nome_genero")
)
df_dim_genres.write.format("delta").mode("overwrite").saveAsTable("gold.dim_genres")

df_bridge_genre = (
    df_gen.join(df_movies_lookup, on="id_filme", how="inner")
    .join(spark.table("gold.dim_genres"), on="nome_genero", how="inner")
    .select("sk_movie_id", "sk_genre_id")
    .distinct()
)
df_bridge_genre.write.format("delta").mode("overwrite").saveAsTable("gold.bridge_movie_genre")
```

### 2.3 Dimensão Pessoas, Empresas e Bridges

Cria as dimensões de pessoas e empresas, com suas respectivas tabelas bridge.

```python
df_pe = spark.table("silver.tb_pessoas_empresas")

df_dim_people = (
    df_pe.filter(col("tipo_entidade").isin("Ator", "Diretor", "Roteirista"))
    .select(col("nome_entidade").alias("nome_pessoa"), col("tipo_entidade").alias("tipo_pessoa"))
    .distinct()
    .withColumn("sk_person_id", monotonically_increasing_id().cast(LongType()))
    .select("sk_person_id", "nome_pessoa", "tipo_pessoa")
)
df_dim_people.write.format("delta").mode("overwrite").saveAsTable("gold.dim_people")

df_bridge_person = (
    df_pe.filter(col("tipo_entidade").isin("Ator", "Diretor", "Roteirista"))
    .join(df_movies_lookup, on="id_filme", how="inner")
    .join(spark.table("gold.dim_people"), 
          (col("nome_entidade") == col("nome_pessoa")) & (col("tipo_entidade") == col("tipo_pessoa")), 
          how="inner")
    .select("sk_movie_id", "sk_person_id")
    .distinct()
)
df_bridge_person.write.format("delta").mode("overwrite").saveAsTable("gold.bridge_movie_person")

df_dim_companies = (
    df_pe.filter(col("tipo_entidade") == "Produtora")
    .select(col("nome_entidade").alias("nome_produtora"))
    .distinct()
    .withColumn("sk_company_id", monotonically_increasing_id().cast(LongType()))
    .select("sk_company_id", "nome_produtora")
)
df_dim_companies.write.format("delta").mode("overwrite").saveAsTable("gold.dim_companies")

df_bridge_company = (
    df_pe.filter(col("tipo_entidade") == "Produtora")
    .join(df_movies_lookup, on="id_filme", how="inner")
    .join(spark.table("gold.dim_companies"), col("nome_entidade") == col("nome_produtora"), how="inner")
    .select("sk_movie_id", "sk_company_id")
    .distinct()
)
df_bridge_company.write.format("delta").mode("overwrite").saveAsTable("gold.bridge_movie_company")
```

### 2.4 Dimensão Reviews

Agrega avaliações de usuários por filme: quantidade e nota média.

```python
df_reviews = spark.table("silver.tb_avaliacoes_usuarios")

df_dim_reviews = (
    df_reviews.groupBy("id_filme")
    .agg(
        count("*").cast(IntegerType()).alias("qtd_avaliacoes_usuarios"),
        spark_round(avg("nota_usuario"), 2).cast(DoubleType()).alias("nota_media_usuarios")
    )
    .join(df_movies_lookup, on="id_filme", how="inner")
    .withColumn("sk_review_id", monotonically_increasing_id().cast(LongType()))
    .select("sk_review_id", "sk_movie_id", "qtd_avaliacoes_usuarios", "nota_media_usuarios")
)
df_dim_reviews.write.format("delta").mode("overwrite").saveAsTable("gold.dim_reviews")
```

## 3. Tabela Fato

Junta dados financeiros e métricas de engajamento à dimensão de filmes, deduplicando por `id_filme` para garantir 1:1 com `dim_movies`.

```python
df_fin = spark.table("silver.tb_financeiro_filmes").dropDuplicates(["id_filme"])
df_met = spark.table("silver.tb_metricas_engajamento").dropDuplicates(["id_filme"])

df_fact = (
    df_movies_lookup
    .join(df_fin, on="id_filme", how="left")
    .join(df_met, on="id_filme", how="left")
    .select(
        col("sk_movie_id"),
        col("orcamento_usd").cast(DecimalType(18, 2)),
        col("receita_usd").cast(DecimalType(18, 2)),
        col("lucro_usd").cast(DecimalType(18, 2)),
        col("orcamento_brl").cast(DecimalType(18, 2)),
        col("receita_brl").cast(DecimalType(18, 2)),
        col("lucro_brl").cast(DecimalType(18, 2)),
        col("popularidade").cast(DoubleType()),
        col("nota_media_tmdb").cast(DoubleType()),
        col("qtd_votos_tmdb").cast(IntegerType()),
        col("nota_media_imdb").cast(DoubleType()),
        col("qtd_votos_imdb").cast(IntegerType())
    )
)
df_fact.write.format("delta").mode("overwrite").saveAsTable("gold.fact_movies_performance")
```

## 4. Desafio

### Pergunta 1: Receita total em R$

Soma de `receita_brl` de todos os filmes.

```python
q1 = spark.sql("""
    SELECT 
        SUM(receita_brl) as receita_total_brl
    FROM gold.fact_movies_performance
""")
display(q1)
```

### Pergunta 2: Top 5 filmes por popularidade

Ranking dos 5 filmes com maior valor de popularidade.

```python
q2 = spark.sql("""
    SELECT 
        m.titulo,
        f.popularidade
    FROM gold.fact_movies_performance f
    INNER JOIN gold.dim_movies m ON f.sk_movie_id = m.sk_movie_id
    WHERE f.popularidade IS NOT NULL
    ORDER BY f.popularidade DESC
    LIMIT 5
""")
display(q2)
```

### Pergunta 3: Filmes por gênero

Contagem de filmes distintos por gênero, do maior para o menor volume.

```python
q3 = spark.sql("""
    SELECT 
        g.nome_genero,
        COUNT(DISTINCT b.sk_movie_id) as total_filmes
    FROM gold.dim_genres g
    INNER JOIN gold.bridge_movie_genre b ON g.sk_genre_id = b.sk_genre_id
    GROUP BY g.nome_genero
    ORDER BY total_filmes DESC
""")
display(q3)
```

### Pergunta 4: Top 10 filmes por receita

Os 10 filmes de maior receita com posição no ranking.

```python
q4 = spark.sql("""
    WITH ranking_receita AS (
        SELECT 
            m.titulo,
            f.receita_usd,
            f.receita_brl,
            RANK() OVER (ORDER BY f.receita_usd DESC) as posicao_ranking
        FROM gold.fact_movies_performance f
        INNER JOIN gold.dim_movies m ON f.sk_movie_id = m.sk_movie_id
        WHERE f.receita_usd IS NOT NULL
    )
    SELECT * FROM ranking_receita
    WHERE posicao_ranking <= 10
    ORDER BY posicao_ranking ASC
""")
display(q4)
```

### Pergunta 5: Ator com mais participações nos últimos 2 anos

Considera a data de lançamento mais recente como referência e conta participações de atores nos 24 meses anteriores.

```python
q5 = spark.sql("""
    WITH data_referencia AS (
        SELECT MAX(data_lancamento) as max_data
        FROM gold.dim_movies
        WHERE data_lancamento <= CURRENT_DATE()
          AND status_filme = 'Lançado'
    ),
    filmes_2_anos AS (
        SELECT m.sk_movie_id
        FROM gold.dim_movies m
        CROSS JOIN data_referencia r
        WHERE m.data_lancamento BETWEEN add_months(r.max_data, -24) AND r.max_data
    )
    SELECT 
        p.nome_pessoa as ator,
        COUNT(b.sk_movie_id) as total_participacoes
    FROM gold.dim_people p
    INNER JOIN gold.bridge_movie_person b ON p.sk_person_id = b.sk_person_id
    INNER JOIN filmes_2_anos f ON b.sk_movie_id = f.sk_movie_id
    WHERE p.tipo_pessoa = 'Ator'
    GROUP BY p.nome_pessoa
    ORDER BY total_participacoes DESC
    LIMIT 1
""")
display(q5)
```

### Pergunta 6: Produtora com maior lucro nos últimos 5 anos

Soma o lucro por produtora nos 60 meses anteriores a data de lançamento mais recente.

```python
q6 = spark.sql("""
    WITH data_referencia AS (
        SELECT MAX(data_lancamento) as max_data
        FROM gold.dim_movies
        WHERE data_lancamento <= CURRENT_DATE()
          AND status_filme = 'Lançado'
    ),
    filmes_5_anos AS (
        SELECT 
            m.sk_movie_id,
            f.lucro_usd,
            f.lucro_brl
        FROM gold.dim_movies m
        CROSS JOIN data_referencia r
        INNER JOIN gold.fact_movies_performance f ON m.sk_movie_id = f.sk_movie_id
        WHERE m.data_lancamento BETWEEN add_months(r.max_data, -60) AND r.max_data
          AND f.lucro_usd IS NOT NULL
    )
    SELECT 
        c.nome_produtora,
        ROUND(SUM(f.lucro_usd), 2) as lucro_total_usd,
        ROUND(SUM(f.lucro_brl), 2) as lucro_total_brl
    FROM gold.dim_companies c
    INNER JOIN gold.bridge_movie_company b ON c.sk_company_id = b.sk_company_id
    INNER JOIN filmes_5_anos f ON b.sk_movie_id = f.sk_movie_id
    GROUP BY c.nome_produtora
    ORDER BY lucro_total_usd DESC
    LIMIT 1
""")
display(q6)
```
